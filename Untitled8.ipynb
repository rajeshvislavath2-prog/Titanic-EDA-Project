```python

import pandas as pd
import numpy as np
from sklearn.decomposition import TruncatedSVD
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
movies = pd.DataFrame({
    "movieId": [1, 2, 3, 4, 5, 6, 7, 8],
    "title": [
        "The Matrix", "Inception", "Titanic", "The Notebook",
        "John Wick", "Mad Max: Fury Road", "La La Land", "The Fault in Our Stars"
    ],
    "genres": [
        "Action|Sci-Fi", "Action|Sci-Fi|Thriller", "Romance|Drama", "Romance|Drama",
        "Action|Thriller", "Action|Sci-Fi", "Romance|Musical", "Romance|Drama"
    ]
})
 
ratings = pd.DataFrame({
    "userId":  [1, 1, 1, 2, 2, 3, 3, 3, 4, 4, 4, 5, 5],
    "movieId": [1, 2, 5, 3, 4, 1, 2, 6, 3, 4, 7, 5, 6],
    "rating":  [5, 4, 5, 5, 4, 4, 5, 5, 4, 5, 3, 5, 4]
})
```

```python

user_item_matrix = ratings.pivot_table(index="userId", columns="movieId", values="rating").fillna(0)
```

```python
n_components = min(3, user_item_matrix.shape[1] - 1)  # keep small since sample data is tiny
svd = TruncatedSVD(n_components=n_components, random_state=42)
latent_matrix = svd.fit_transform(user_item_matrix)          # users -> latent space
reconstructed = np.dot(latent_matrix, svd.components_)        # back to movie space (predicted ratings)
 
predicted_ratings = pd.DataFrame(
    reconstructed, index=user_item_matrix.index, columns=user_item_matrix.columns
)
```

```python
def recommend_movies(user_id, top_n=3):
    already_rated = ratings.loc[ratings.userId == user_id, "movieId"].tolist()
    scores = predicted_ratings.loc[user_id].drop(labels=already_rated, errors="ignore")
    top_movie_ids = scores.sort_values(ascending=False).head(top_n).index
    return movies[movies.movieId.isin(top_movie_ids)][["title", "genres"]]
```

```python
tfidf = TfidfVectorizer(token_pattern=r"[^|]+")
genre_matrix = tfidf.fit_transform(movies["genres"])
content_similarity = cosine_similarity(genre_matrix)
 
def similar_movies(title, top_n=3):
    idx = movies[movies.title == title].index[0]
    scores = list(enumerate(content_similarity[idx]))
    scores = sorted(scores, key=lambda x: x[1], reverse=True)[1:top_n + 1]
    return movies.iloc[[i for i, _ in scores]][["title", "genres"]]
```

```python
if __name__ == "__main__":
    print("Top picks for User 1 (collaborative filtering / ML):")
    print(recommend_movies(user_id=1, top_n=3), "\n")
 
    print("Movies similar to 'Inception' (content-based fallback):")
    print(similar_movies("Inception", top_n=3))
```
